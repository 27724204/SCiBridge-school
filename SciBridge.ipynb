# Astronomical Observatories and the Electromagnetic Spectrum

Modern astronomy relies on a global network of observatories located both on the Earth's surface and in space. Each telescope is designed to observe a specific region of the electromagnetic spectrum and is optimised according to its scientific objectives. The characteristics of an instrument, such as its wavelength range, sensitivity, angular resolution, and field of view, determine the types of astronomical objects and physical processes that it can investigate.

The **electromagnetic spectrum** provides the fundamental framework for classifying astronomical telescopes. Since the Earth's atmosphere absorbs many wavelengths of electromagnetic radiation, the location and design of an observatory depend on the radiation it is intended to detect. For example, radio and optical telescopes can operate from the ground, whereas X-ray and gamma-ray telescopes must be placed in space to avoid atmospheric absorption.

## Why Multi-wavelength Astronomy?

No single telescope can provide a complete picture of an astronomical object. Observing the Universe across multiple wavelengths enables astronomers to identify different physical components, including:

- Cold molecular gas (radio)
- Dust clouds (infrared)
- Stars (optical)
- Hot ionised gas (ultraviolet and X-rays)
- Relativistic particles and energetic phenomena (gamma rays)

Combining observations from multiple wavelengths provides a more comprehensive understanding of the physical processes occurring within stars, galaxies, and active galactic nuclei.

### Simple Linear Regression Model

The mathematical model for a simple linear regression line is expressed as:

$$\hat{y} = b_0 + b_1x$$

### Components Breakdown

* **$\hat{y}$** (Predicted Value): The estimated value of the dependent variable.
* **$x$** (Independent Variable): The predictor variable used to estimate the outcome.
* **$b_1$** (Slope): The expected change in $\hat{y}$ for every one-unit increase in $x$.
* **$b_0$** (Y-intercept): The value of $\hat{y}$ when $x$ is equal to zero.

In Python is [linear regression model](https://docs.scipy.org/doc//scipy/reference/generated/scipy.stats.mstats.linregress.html) *


```python
from scipy.stats import linregress 
```

For compatibility with older versions of SciPy, the return value acts like a named tuple of length 5, with fields slope, intercept, rvalue, pvalue and stderr, so one can continue to write:

```python
#Converting the time to second such that we get it in float64
voyage2_cleaned_data["observation_T_sec"]= (
    voyage2_cleaned_data["observation_T"]
    - voyage2_cleaned_data["observation_T"].iloc[0]
).dt.total_seconds()
```

```python
voyage2_cleaned_data = voyage2_cleaned_data.copy()

voyage2_cleaned_data["observation_T_sec"] = (
    voyage2_cleaned_data["observation_T"]
    - voyage2_cleaned_data["observation_T"].iloc[0]
).dt.total_seconds()
```

```python
#linear regression 
slope, intercept, r, p, se = linregress(voyage2_cleaned_data['observation_T_sec'], voyage2_cleaned_data['Radial Distance (AU)'])
```
